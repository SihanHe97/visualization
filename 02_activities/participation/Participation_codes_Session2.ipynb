```python
# ensuring the reproduciblity of the data/figure
np.random.seed(613)
x = np.arange(50)
y1 = np.random.randint(0,100,50)
y2 = np.random.randint(0,100,50)
```

Activity: Multiple y variables on the same axes
We created a figure using plt.subpolots function, and plotted multiple y variables using ax.plot against the same x value.

```python
# Activity: Multiple y variables on the same axes
fig, ax = plt.subplots(figsize=(5, 3))
ax.plot(x,y1)
ax.plot(x,y2)
fig.show()
```

```python
# add legend
fig, ax = plt.subplots(figsize=(5, 3))
ax.plot(x,y1, label = "Person 1" )
ax.plot(x,y2, label = "Person 2" )
ax.legend(loc='upper right')
```

```python
# modifing legends
fig, ax = plt.subplots(figsize=(5, 3))
ax.plot(x,y1, label = "Person 1")
ax.plot(x,y2, label = "Person 2")
ax.legend(loc='upper center',
        frameon = True,
        fontsize = 12,
        ncol = 2,
        shadow = True )
```

```python
#moving legend outside the graph
fig, ax = plt.subplots(figsize=(5, 3))
ax.plot(x,y1, label = "Person 1")
ax.plot(x,y2, label = "Person 2")
ax.legend(loc='upper left',
        bbox_to_anchor =(0.7, 1.3))
fig.show()
```

```python
# adding annotations

# create a scatter plot
fig, ax = plt.subplots(figsize=(5, 3))
ax.scatter(x,y1, label = "Person 1")
ax.scatter(x,y2, label = "Person 2")
ax.legend(loc='lower right', prop = {"family":"serif"})

# including texts
ax.text(10, 110, "This figure is important!")

# modifying the text
ax.text(70, 110, "Is this figure important?",
      ha='right',
      color = 'red', 
      size = 10)
fig.show()
```

```python
# positioning the text
# making an empty set
fig, ax = plt.subplots()
ax.axis([0, 10, 0, 10])

# positioning
ax.text(1, 5, ". Data:(1, 5)",           transform=ax.transData    )
ax.text(0.5, 0.1, ". Axes:(0.5, 0.1)",   transform=ax.transAxes    )
ax.text(0.6, 0.8, ". Figure:(0.6, 0.8)", transform=fig.transFigure )
```

```python
# including the arrows
fig, ax = plt.subplots(figsize=(5, 3))
ax.scatter(x,y1, label = "Person 1")
ax.scatter(x,y2, label = "Person 2")
ax.legend(loc='lower right', prop = {"family":"serif"})
ax.annotate('This is important!', xy=(10, 95), xytext=(30, 94),
          arrowprops=dict(facecolor='black'))
fig.show()
```

```python
# modifying the arrows
# <->
fig, ax = plt.subplots(figsize=(5, 3))
ax.scatter(x,y1, label = "Person 1")
ax.scatter(x,y2, label = "Person 2")
ax.legend(loc='lower right', prop = {"family":"serif"})
ax.annotate('This is important!',
          xy=(10, 95), xytext=(20, 94),
          arrowprops = dict(arrowstyle = "<->",
                            color = "hotpink"))
fig.show()
```

```python
# modifying the arrows
# -[
fig, ax = plt.subplots(figsize=(5, 3))
ax.scatter(x,y1, label = "Person 1")
ax.scatter(x,y2, label = "Person 2")
ax.legend(loc='lower right', prop = {"family":"serif"})
ax.annotate('This is important!',
          xy=(10, 95), xytext=(20, 94),
          arrowprops = dict(arrowstyle = "-[",
                            color = "hotpink"))
fig.show()
```

```python
# modifying the arrows
# fancy
fig, ax = plt.subplots(figsize=(5, 3))
ax.scatter(x,y1, label = "Person 1")
ax.scatter(x,y2, label = "Person 2")
ax.legend(loc='lower right', prop = {"family":"serif"})
ax.annotate('This is important!',
          xy=(10, 95), xytext=(20, 94),
          arrowprops = dict(arrowstyle = "fancy",
                            color = "green"))
fig.show()
```

```python
# non-locator--removing ticks and labels
fig, ax = plt.subplots(figsize=(5, 3))
ax.scatter(x,y1, label = "Person 1")
ax.scatter(x,y2, label = "Person 2")
ax.legend(loc='lower right')

ax.yaxis.set_major_locator(plt.NullLocator()) # remove y ticks
ax.xaxis.set_major_locator(plt.NullLocator()) # remove x ticks

ax.yaxis.set_major_formatter(plt.NullFormatter()) # remove y labels
ax.xaxis.set_major_formatter(plt.NullFormatter()) # remove x labels
```

```python
# limiting locator
fig, ax = plt.subplots(figsize=(5, 3))
ax.scatter(x,y1, label = "Person 1")
ax.scatter(x,y2, label = "Person 2")
ax.legend(loc='lower right')
ax.xaxis.set_major_locator(plt.MaxNLocator(8))
```

```python
# multiple locators
fig, ax = plt.subplots(figsize=(5, 3))
ax.scatter(x,y1, label = "Person 1")
ax.scatter(x,y2, label = "Person 2")
ax.legend(loc='lower right')
ax.xaxis.set_major_locator(plt.MultipleLocator(5))
```

```python
# rotating labels
fig, ax = plt.subplots(figsize=(5, 3))
ax.scatter(x,y1, label = "Person 1")
ax.scatter(x,y2, label = "Person 2")
ax.legend(loc='lower right')
plt.xticks(rotation=45, ha='right')
```
