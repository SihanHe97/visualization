Activity: Modifying axis title font
Adding a label with plt.xlabel, and modify this with fontdict. Using ax.set_xlabel, though, allows us to add label/title to a specific axis (could be helpful when having multiple subplots).

```python
# Producing the data from last session
np.random.seed(613)
x = np.arange(50)
y1 = np.random.randint(0,100,50)
y2 = np.random.randint(0,100,50)
```

```python
# activity modifying axis title font
font1 = {'family':'serif','color':'indigo'}
fig, ax = plt.subplots(figsize=(5, 3))
ax.scatter(x,y1, label = "Person 1")
ax.scatter(x,y2, label = "Person 2")
ax.legend(loc='lower right')
ax.set_xlabel('Shiny New X Axis!', fontsize = 18, fontdict = font1)
#plt.xlabel('Shiny New X Axis!', fontsize = 18, fontdict = font1)
```

```python
# setting up the data for this session
np.random.seed(613)
x1 = np.arange(50)
y1 = np.random.randint(0, 75,50)
x2 = np.array(["Luffy", "Zoro", "Nami", "Usopp", "Sanji"])
y2 = np.array([110, 180, 240, 99, 220])
```

```python
# defining multiple subplots. One scatter and the other bar graph
fig, (ax1, ax2) = plt.subplots(ncols=2,
                        nrows=1,
                        figsize=(7, 3))
ax1.scatter(x1,y1)
ax2.bar(x2,y2)
```

Activity Customizing the plots.


```python
#
fig, (ax1, ax2) = plt.subplots(ncols=2,
                        nrows=1,
                        figsize=(7, 3))
ax1.scatter(x1,y1)
ax1.annotate('This is important!',
          xy=(10, 65), xytext=(20, 80),
          arrowprops = dict(arrowstyle = "<->",
                            color = "green"))
ax2.bar(x2,y2)
```

```python
# arrange the subplots
fig, someaxes = plt.subplot_mosaic([['ax1', 'ax3'],
                                ['ax2', 'ax3']],
                                figsize=(8, 4))
someaxes["ax1"].scatter(x1,y1)
someaxes["ax2"].bar(x2,y2)
someaxes["ax3"].plot(x1,y1)
```

```python
# modifying the layout
fig, someaxes = plt.subplot_mosaic([['ax1', 'ax3'],
                                ['ax2', 'ax3']],
                                figsize=(8, 4)) # Default layout cut off one of the x titles
someaxes["ax1"].scatter(x1,y1)
someaxes["ax2"].bar(x2,y2)
someaxes["ax3"].plot(x1,y1)
someaxes["ax1"].set_xlabel('A Big Label',fontsize=18)
someaxes["ax2"].set_xlabel('Another Label',fontsize=18)
someaxes["ax3"].set_xlabel('Label 2: 2 Fast 2 Furious',fontsize=18 )
```

```python
# adjusting layout
fig, someaxes = plt.subplot_mosaic([['ax1', 'ax3'],
                                ['ax2', 'ax3']],
                                figsize=(8, 4),
                                layout = 'constrained')
someaxes["ax1"].scatter(x1,y1)
someaxes["ax2"].bar(x2,y2)
someaxes["ax3"].plot(x1,y1)
someaxes["ax1"].set_xlabel('A Big Label',fontsize=18)
someaxes["ax2"].set_xlabel('Another Label',fontsize=18)
someaxes["ax3"].set_xlabel('Label 2: 2 Fast 2 Furious',fontsize=18 )
#fig.tight_layout()
```

```python
# multiple visulizations on one axis
# first make our sample data
x = np.array(["Luffy", "Zoro", "Nami", "Usopp", "Sanji"])
y1 = np.array([110, 180, 240, 99, 220])
y2 = np.array([170, 100, 90, 120, 50])
# define our figure and axes (just one this time)
fig, ax = plt.subplots(figsize=(8, 3))
# now call both bar and plot elements to the same axes (ax)
ax.bar(x, y1,
color = "blue")
ax.plot(x, y2,
color = "hotpink")
```

```python
# Add error information
# calculate standard deviation
y2_sd = np.std(y2)

# plot the previous line
fig, ax = plt.subplots(figsize=(8, 3))
ax.plot(x, y2, color = "red")

# add error bars
ax.errorbar(x,
        y2,
        color = 'black',
        yerr = y2_sd,
        fmt = "none")
```

```python
# customize error bars
fig, ax = plt.subplots(figsize=(8, 3))
ax.plot(x, y2,
        color = "blue")
ax.errorbar(x,
        y2,
        yerr = y2_sd,
        fmt = "none",
        ecolor= "black",
        elinewidth= 4,
        capsize = 6,
        capthick= 4 )
```

```python
# using Errorevery
y2_sd = np.std(y2)

# plot the previous line
fig, ax = plt.subplots(figsize=(8, 3))
ax.plot(x, y2,
        color = "blue")
ax.errorbar(x,
        y2,
        yerr = y2_sd,
        fmt = "none",
        ecolor = "indigo",
        elinewidth = 4,
        capsize = 6,
        capthick = 4,
        errorevery= 2 )
```
